```python
import pandas as pd
import numpy as np

# 1. Información de los 12 Grupos extraída del PDF
groups_data = [
    ("Mexico", "A"), ("Korea Republic", "A"), ("South Africa", "A"), ("Czechia", "A"),
    ("Canada", "B"), ("Switzerland", "B"), ("Qatar", "B"), ("Bosnia and Herzegovina", "B"),
    ("Brazil", "C"), ("Morocco", "C"), ("Scotland", "C"), ("Haiti", "C"),
    ("United States", "D"), ("Paraguay", "D"), ("Australia", "D"), ("Turkiye", "D"),
    ("Germany", "E"), ("Ecuador", "E"), ("Cote d'Ivoire", "E"), ("Curacao", "E"),
    ("Netherlands", "F"), ("Japan", "F"), ("Sweden", "F"), ("Tunisia", "F"),
    ("Belgium", "G"), ("IR Iran", "G"), ("Egypt", "G"), ("New Zealand", "G"),
    ("Spain", "H"), ("Uruguay", "H"), ("Saudi Arabia", "H"), ("Cape Verde", "H"),
    ("France", "I"), ("Senegal", "I"), ("Norway", "I"), ("Iraq", "I"),
    ("Argentina", "J"), ("Austria", "J"), ("Algeria", "J"), ("Jordan", "J"),
    ("Portugal", "K"), ("Colombia", "K"), ("Uzbekistan", "K"), ("Congo DR", "K"),
    ("England", "L"), ("Croatia", "L"), ("Ghana", "L"), ("Panama", "L")
]
df_groups = pd.DataFrame(groups_data, columns=["equipo", "grupo_pdf"])

# 2. Cuotas y Probabilidades Implícitas (Mercado de Apuestas - Mayo 2026)
odds_data = [
    ("Spain", 475, 400, 475, 17.4),
    ("France", 500, 700, 500, 16.7),
    ("England", 650, 600, 650, 13.3),
    ("Brazil", 800, 800, 850, 10.5),
    ("Argentina", 900, 900, 900, 9.0),
    ("Portugal", 1000, 1100, 1000, 9.0),
    ("Germany", 1400, 1200, 1400, 6.7),
    ("Netherlands", 2000, 1600, 2200, 4.8),
    ("Norway", 3000, 3500, 3500, 2.9),
    ("Belgium", 3500, 2500, 3500, 2.9),
    ("Colombia", 4000, 5000, 5000, 2.4),
    ("Uruguay", 5000, 6000, 6000, 1.9),
    ("United States", 6000, 7000, 6500, 1.5),
    ("Mexico", 7000, 7000, 7000, 1.4),
    ("Japan", 6500, 7500, 6500, 1.4),
    ("Morocco", 7500, 8000, 7500, 1.3)
]
df_odds = pd.DataFrame(odds_data, columns=["equipo", "draftkings_odds", "betmgm_odds", "fox_dk_odds", "prob_implicita_pct"])

# 3. Historial en Copas del Mundo
hist_data = [
    ("Brazil", 5, 8, 11, 2002),
    ("Germany", 4, 8, 13, 2014), 
    ("Argentina", 3, 6, 8, 2022),
    ("France", 2, 3, 5, 2018),
    ("Spain", 1, 1, 3, 2010),
    ("England", 1, 1, 4, 1966),
    ("Portugal", 0, 0, 1, np.nan),
    ("Netherlands", 0, 3, 4, np.nan)
]
df_hist = pd.DataFrame(hist_data, columns=["equipo", "titulos_historicos", "finales_historicas", "semis_historicas", "ultimo_titulo"])

# 4. Consolidar Base de Contexto (PDF)
df_contexto = df_groups.merge(df_odds, on="equipo", how="left").merge(df_hist, on="equipo", how="left")
# Rellenar con ceros a los equipos sin historial o sin cuotas publicadas
df_contexto.fillna({
    "draftkings_odds": 10000, "betmgm_odds": 10000, "fox_dk_odds": 10000, 
    "prob_implicita_pct": 0.5, "titulos_historicos": 0, "finales_historicas": 0, 
    "semis_historicas": 0
}, inplace=True)

# 5. Cargar los Match Logs extraídos por tu scraper
df_matchlogs = pd.read_csv("fbref_matchlogs.csv")

# 6. FUSIÓN MAESTRA (Merge)
df_master = df_matchlogs.merge(df_contexto, on="equipo", how="left")

# Exportar Dataset Final para el Modelo de ML
df_master.to_csv("world_cup_ml_dataset.csv", index=False)
print(f"✅ ¡Dataset fusionado con éxito! {df_master.shape[0]} filas listas para entrenar.")
```
